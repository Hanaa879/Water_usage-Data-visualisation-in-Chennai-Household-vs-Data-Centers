```python
import pandas as pd
```

```python
#we import the excel data file that we saved as csv
df = pd.read_csv('Water_data.csv')
```

```python
# Define the annual water need for 1 Chennai family (135L * 4 people * 365 days)
family_need = 197100
```

```python
# Extract Microsoft's total from your sheet
ms_total = df.loc[df['Entity'] == 'Microsoft Global', 'Liters_Per_Year'].values[0]
```

```python
# The "Data Story" Math = Relatability conversion which translates large numbers to tangible values to conceptualise
num_families = ms_total / family_need
```

```python
print(f"One year of Microsoft's water use could support {int(num_families):,} Chennai families.")
```
